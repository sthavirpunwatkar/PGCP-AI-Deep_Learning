```python
# Topic 3: Language as a Rule-Governed Dynamic System.
# structural rules (syntax)


# subject and predicates
def check_agreement(subject, verb):
  rules = {
      "singular_subjects": ["he", "she", "the student"],
      "singular_verb": ["studies", "learns"]
  }

  if subject in rules["singular_subjects"] and verb in rules["singular_verb"]:
    return "Valid Sentence Structure"   # True
  else:
    return "Structural Rule Violation"    # Error

print(check_agreement("the student", "studies"))
print(check_agreement("he", "learns"))
print(check_agreement("the student", "studies studies"))
```

```python
# 3B
# Implemeting an Out-of-Vocabulary (OOV) Fallback Rule
# Parser: programmatic check module
def resilient_dymamic_parser(user_phrase):
  known_vocabulary = ["scholar", "studies", "computer", "networks"]
  input_tokens = user_phrase.lower().split()      # .lower: BIGIT IS GOOD -> bigit is good


  processed_output = []
  for token in input_tokens:
    if token in known_vocabulary:
      processed_output.append(token)

    else:
      processed_output.append(f"[OOV_TAG: '{token}']")

  return processed_output

test_phrase = "The scholar prompt-engineers computer networks"
print("Dymaic Resilient Output: \n", resilient_dymamic_parser(test_phrase))
```

```python
# 5 Language as a Biological, Social, and Phychological Phenomemon
def psychological_parser(sentence):
    words = sentence.split()

    # Check for consecutive identical words (stuttering or typographical loop)
    duplicate_count = 0
    for i in range(len(words) - 1):   # 6 - 1 = 5
        if words[i] == words[i+1]:
            duplicate_count += 1

    if duplicate_count > 1:
        return f"⚠️ Cognitive Distraction! Found {duplicate_count} consecutive repeating word sequences. Processing slowed."
    return "✅ Text processed comfortably by cognitive working memory."

print(psychological_parser("The scholar analyzes data data carefully."))  # one duplicate
print(psychological_parser("The scholar parsed the the data set very very quickly.")) # two duplicates
```

```python
# Character-level Noise simulation
# Cell 7B: Simulating Typing Noise in Written Communication
import random

def inject_typographical_noise(clean_text, noise_probability=0.3):
    chars = list(clean_text)
    for i in range(len(chars)):
        # Randomly decide whether to corrupt a character boundary
        if random.random() < noise_probability and chars[i].isalpha():
            chars[i] = 'X' # Substitute character with noise symbol
    return "".join(chars)

original_written = "computer"
print(f"📝 Original Written Form: {original_written}")
print(f"💥 Noisy Transmission Form: {inject_typographical_noise(original_written)}")
```

## 1B. Form, Function and Meaning in Language Analysis

1. Form: What the toekn looks like or its raw spelling configuration
2. Function: The specific structral job that token performs inside the sentence layout.
3. Meaning : The specific conceptual asset or literal idea the word communicates.

```python
# Cell 3: Mapping Form, Function, and Meaning Explicitly
class LinguisticTokenProperties:
    def __init__(self, target_form, syntactic_function, semantic_meaning):
        self.form = target_form                 # Physical spelling representation
        self.function = syntactic_function       # The structural job in the layout
        self.meaning = semantic_meaning         # The core literal definition concept

# Instantiating structural instances to track individual token configurations
token_analysis_1 = LinguisticTokenProperties(
    target_form="scholar",
    syntactic_function="Noun Subject (nsubj)",
    semantic_meaning="An individual engaged in advanced systematic research"
)

token_analysis_2 = LinguisticTokenProperties(
    target_form="studies",
    syntactic_function="Root Action Verb (ROOT)",
    semantic_meaning="The active execution of analytical investigation"
)

# Displaying properties cleanly
all_tokens_metadata = [token_analysis_1, token_analysis_2]
for idx, token in enumerate(all_tokens_metadata):
    print(f"📊 Analyzing Token Property Matrix #{idx+1}:")
    print(f"  * [Form Layer]     What it looks like : '{token.form}'")
    print(f"  * [Function Layer] What its job is     : {token.function}")
    print(f"  * [Meaning Layer]  What it communicates: {token.meaning}\n")
```

```python
# Cell 5: Advanced Pragmatic Context Evaluation Simulation
def evaluate_pragmatic_subtext(input_sentence, external_environment_context):
    # Literal Semantic Layer Interpretation
    base_semantic_meaning = "Literal Statement: The temperature of this room is lower than comfortable bounds."

    # Pragmatic Layer Check: Evaluating real-world situational parameters
    speaker_profile = external_environment_context.get("speaker_role")
    room_condition = external_environment_context.get("room_physical_state")
    audience_present = external_environment_context.get("is_student_present")

    # Contextual inference decision logic
    if speaker_profile == "professor" and room_condition == "freezing" and audience_present == True:
        pragmatic_intent = "⚠️ Inferred Contextual Intention: 'Please go and close the window immediately!'"
    elif speaker_profile == "scholar" and room_condition == "freezing":
        pragmatic_intent = "⚠️ Inferred Contextual Intention: 'I should turn on the laboratory heating unit.'"
    else:
        pragmatic_intent = "Standard Alignment: No urgent hidden action or subtext inferred."

    return base_semantic_meaning, pragmatic_intent

# Creating explicit real-world context parameters outside the raw text input string
active_laboratory_context = {
    "speaker_role": "professor",
    "room_physical_state": "freezing",
    "is_student_present": True
}

literal_text = "It is cold in here."
semantic_layer, pragmatic_layer = evaluate_pragmatic_subtext(literal_text, active_laboratory_context)

print("📝 Raw String Input Statement:", literal_text)
print("🧬 [Semantic Level Interpretation] :", semantic_layer)
print("🧠 [Pragmatic Level Context Shift] :", pragmatic_layer)
```

## 3A. Natural Lanuage Understanding (NLU)

NLU is a specific AI sub-discipline focused on reading raw human text and convering it into highly organizes, structured programmatic data array( like intent classification or variable records)

```python
def nlu_intent_extractor(user_command_string):
  normalized_input = user_command_string.lower().strip()

  has_save_action = any(verb in normalized_input for verb in ["save", "backup", "export", "archieve"])
  has_data_target = any(noun in normalized_input for noun in ["data", "results", "matrix", "scholar"])

  # x = 0
  # Name:
  # Rollno:
  extracted_structured_data ={
      "detected_intent": "UNKNOWN",
      "parameters": {"target_type": None, "priority_high": False}   # False = 0, True = 1
  }

  if has_save_action and has_data_target:
    extracted_structured_data["detected_intent"] = "EXECUTE_SYSTEM_BACKUP"
    extracted_structured_data["parameters"]["target_type"] = "research_database"
    if "urgent" in normalized_input or "immediately" in normalized_input:
      extracted_structured_data["parameters"]["priority_high"] = True

  return extracted_structured_data

input = "Please backup the scholar results immediately!"
nlu_output_record = nlu_intent_extractor(input)

print("Raw input string: ", input)

print("NLU Structured record Object:")
for key, value in nlu_output_record.items():
  print(f"  * Field: {key:<16} -> Structured Value: {value}")
```

NLG: The mathematical inverser of NLU, - taking highly structured machien data objects and translating them into fuild readable human-style sentences.

NLI: The end-to-end operational wrapper loop combining NLU and NLG together to let a human talk back and forth with an AI system dynamically.

```python
# Simulating an Interactive Conversational Loop (NLI System)
def conditional_nlg_response_generator(machine_state_record):
  active_intent = machine_state_record.get("detected_intent")
  state_params = machine_state_record.get("parameters", {})

  # check the intent type
  if active_intent == "EXECUTE_SYSTEM_BACKUP":
    base_sentence = f" AI Response: I have successfully initiated a backup of the {state_params['target_type']}"
    if state_params.get("priority_high") == True:
      base_sentence += " High priority status confirmed-allocating secondary storage clusters"
    return base_sentence

  return "AI Response: Operational parameters unrecognized. system standing by"

simulated_nlu_state = {
    "detected_intent": "EXECUTE_SYSTEM_BACKUP",
    "parameters": {
        "target_type": "research_database",
        "priority_high": True
    }
}

generated_human_speech = conditional_nlg_response_generator(simulated_nlu_state)
print("NLG output sentence string: ", generated_human_speech)

```

## 2. Shallow Parsing and Phrase Chunking

Shallow Parsing and Phrase Chunking
Shallow parsing takes the POS tags, groups adjacent, related words into non-overlapping grammatical phrases called **chunks**,

this completely skps deep hierarchical nested networks

```python
import nltk
from nltk import pos_tag, word_tokenize
from nltk.chunk import RegexpParser

nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')
nltk.download('punkt_tab') # Explicitly download punkt_tab here as requested by error

sentence = "The quick brown fox jumps over the lazy dog."

tokens = word_tokenize(sentence)
pos_tags = pos_tag(tokens)

grammar = "NP: {<DT>?<JJ>*<NN>}"
chunk_parser = RegexpParser(grammar)

result = chunk_parser.parse(pos_tags)

print(sentence)
print(result)
```

Phrase Chunking: Shallow parsing strategy, that clusters individual tokens into small, non-overlapping grammatical phrases based on PoS patterns WITHOUT computing full sentence hierarchies.

## 3B: Universal Networking Language (UNL)
advanced digital meta-language architecture
designed to represent sematnic information extracted from sentences in a completely language-independent format.

it converts language logic into an interconnected relational graph consisting of universal words (UW nodes) and binary links tracking relations (like agent or object)

```python
                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                              # Cell 12: Modeling a Universal Networking Language (UNL) Relational Graph Object
class UNLRelationGraph:
    def __init__(self):
        self.universal_words = []   # nodes
        self.binary_relations = []  # edges

    def add_node(self, concept, attributes):
        self.universal_words.append({"concept": concept, "attr": attributes})

    def add_relation(self, rel_type, src, tgt):
        self.binary_relations.append(f"{rel_type}({src}, {tgt})")   # rel_type = relationship type, src = starting word, tgt = target word

# Translating the sentence: "The scholar analyzes data." into UNL meta-data logic
unl_system = UNLRelationGraph() # initialize
unl_system.add_node("scholar(iof=person)", attributes=["@entry", "@present"])   # iof = instance of
unl_system.add_node("analyze(iof=action)", attributes=["@entry"])
unl_system.add_relation("agt", "analyze", "scholar") # agt = Agent relation #**

print("🌐 Abstract UNL Nodes & Relations:")
print("👉 Node Concept:", unl_system.universal_words[0])
print("👉 Graph Edge:", unl_system.binary_relations[0])
```

```python
# Mimicking a Semantic Parser (Text to Lambda Calculus Query Mapping)
def naive_semantic_parser(query_string):
    tokens = query_string.lower().split()
    if "find" in tokens and "papers" in tokens:
        topic = tokens[tokens.index("on") + 1] if "on" in tokens else "general"
        # Map the flat text string straight to a formal executable query logic format
        return f"λx.Paper(x) ∧ HasTopic(x, '{topic.upper()}')"
    return "λx.Unknown(x)"

text_query = "Find papers on NLP"
logical_form = naive_semantic_parser(text_query) # translates plain text direclty into a formal lambda calculus representaiton statement (λx)

print("📥 Natural Language Input Query:", text_query)
print("⚙️ Executable Semantic Logic Output:", logical_form)
```

##Chomsky Normal Form (CNF)
CFG = (Variable, Terminals, Product rules (R), Start symbol)
Context free grammer R. A -> a

```python
# Cell 1: Building and Validating Chomsky Normal Form (CNF) Production Maps
import nltk

# Define a grammar explicitly adhering to strict binary branching and terminal constraints
cnf_grammar = nltk.CFG.fromstring("""
  S -> NP VP
  VP -> V NP
  NP -> 'John' | 'apples' | 'Sthavir' | 'water' | 'football' |'goal'
  V -> 'eats' | 'drinks' | 'kicked' | 'scores'
""")


print("⚙️ Verifying Strict Chomsky Normal Form Constraints:")
print("-" * 60)
for production in cnf_grammar.productions():    #John as NP1, apples as NP2
  is_cnf = False
  rhs = production.rhs()    #S->NP VP

  #Check Condition 1: A->BC (Exactly for two non-terminals)
  if len(rhs) == 2 and all(nltk.grammar.is_nonterminal(sym) for sym in rhs):
    is_cnf = True

  #Check Condition 2: A->a (Exactly one terminal symbol)
  elif len(rhs) == 1 and isinstance(rhs[0], str):
    is_cnf = True

  print(f"Rule: {str(production):<22} | valid CNF Element: {is_cnf}")
```

## The CYK Algorithm
Cocke-Younger-Kasami (CYK) algorithm - parsing algorithm to determine whether a string belongs to the language created by CFG.

- dynamic programming,
- breaking down a large parsing problem into smaller sub-problems.

```python
import nltk # Required for accessing CFG productions

words_list = ["Sthavir", "drinks", "water"]
n = len(words_list)

# Initialize the CYK table with sets for each cell
# The table size should be n x n
cyk_table = [[set() for _ in range(n)] for _ in range(n)]

# --- Step 1: Fill the diagonal for length 1 substrings (terminals) ---
# For each word in the input sentence
for i, word in enumerate(words_list):
    # Check all productions in the CFG
    for production in cnf_grammar.productions():
        # If a production is of the form A -> 'word'
        if len(production.rhs()) == 1 and production.rhs()[0] == word:
            # Add the non-terminal A to the corresponding cell in the table
            cyk_table[i][i].add(production.lhs().symbol())

# --- Step 2: Fill the rest of the table for lengths from 2 to n ---
# Iterate over substring lengths (l)
for length in range(2, n + 1):
    # Iterate over starting positions (i)
    for i in range(n - length + 1):
        j = i + length - 1 # Calculate the ending position (j)

        # Iterate over all possible split points (k) within the substring
        for k in range(i, j):
            # For each binary production A -> B C in the grammar
            for production in cnf_grammar.productions():
                if len(production.rhs()) == 2: # Ensure it's a binary production
                    B = production.rhs()[0].symbol()
                    C = production.rhs()[1].symbol()

                    # Check if B can derive the left part (from i to k)
                    # and C can derive the right part (from k+1 to j)
                    if B in cyk_table[i][k] and C in cyk_table[k + 1][j]:
                        # If both are true, then A can derive the entire substring
                        cyk_table[i][j].add(production.lhs().symbol())

# --- Step 3: Print the populated CYK table ---
print("--- Populated CYK Table ---")
for i in range(n):
    for j in range(i, n): # Standard representation: j >= i
        substring = words_list[i : j + 1]
        print(f"Cell[{i+1},{j+1}] (Tokens: {substring}) -> Content: {cyk_table[i][j]}")

# --- Check if the entire sentence can be derived from the start symbol S ---
if 'S' in cyk_table[0][n-1]:
    print(f"\nThe sentence '{' '.join(words_list)}' IS accepted by the grammar.")
else:
    print(f"\nThe sentence '{' '.join(words_list)}' IS NOT accepted by the grammar.")
```

## Semantic Role Identification
syntactic tress: NP, V, P.
Semantic roles: mapping

mapping who or what acts as the driving Agent(Performer), vs. What receives the action as the Patient/Object.

```python
# Cell 3: Rule-Driven Mapping of Core Functional Semantic Roles
def extract_semantic_roles(token_tuple_list):
    semantic_role_map = {}

    # Process sequential lexical components to label operational behaviors
    for word, pos, structure in token_tuple_list:
        #check if the dependency relation of current word is  nominal subject
        if structure == "nsubj":
            semantic_role_map["Agent (Performer)"] = word
        #checks if the Part-of-Speech tag (pos) is 'VBZ',
        elif pos == "VBZ":
            semantic_role_map["Action Predicate"] = word
        #hecks if the dependency relation is 'dobj' (direct object)
        elif structure == "dobj":
            semantic_role_map["Patient / Object"] = word

    return semantic_role_map

# Hypothetical feature input stream matching: [Word, POS-Tag, Dependency-Relation]
sentence_stream = [("John", "NNP", "nsubj"), ("eats", "VBZ", "ROOT"), ("apples", "NNS", "dobj")]
print("🎯 Assigned Meaning Representation Layer:")
print("-" * 60)
print(extract_semantic_roles(sentence_stream))
```
